```python
import pandas as pd
import numpy as np
print(f"Numpy: {np.__version__}")
print(f"Pandas: {pd.__version__}")
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
print(f"Размер датасета: {df.shape}")
print(f"Первые строки: {df.head()}")
print(f"Типы данных: {df.dtypes}")
print(f"Основные статистические данные: {df.describe()}")
print(f"Колонки: {df.columns}")
```

```python
#Медиана стоимости билета среди мужчин, которые не выжили
filtered = df[(df['Sex'] == 'male') & (df['Survived'] == 0)]['Fare']
median_fare = filtered.median()
print(f"Медиана стоимости билетов погибших мужчин: {median_fare}")
```

```python
def age_group(age):
    if pd.isna(age): return 'Unknown'
    if age < 18: return 'Child'
    else: return 'Adult'

df['Age_group'] = df['Age'].apply(age_group)
print(df['Age_group'].value_counts())
```

## Вывод
В датасете `titanic` содержится 891 пассажиров. 
Выживаемость зависела от пола и класса: женщины и пассажиры 1-го класса имели значительно более высокие шансы. 
Медиана стоимости билета среди погибших мужчин составила 9.42 USD.
