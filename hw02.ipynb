```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print(f"Pandas: {pd.__version__}")
print(f"Numpy: {np.__version__}")
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
df_first_100 = df.head(100)
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(df_first_100['Fare'], color="royalblue", marker="o", linewidth=2, label="Fare")

ax.set_title("First 100 passengers' tickets fare")
ax.set_xlabel("Passenger id")
ax.set_ylabel("Fare")
ax.legend()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.hist(df['Age'], color="tab:purple", edgecolor="black", bins=40)

ax.set_title("Passengers' age histogramm")
ax.set_ylabel("Amount of people")
ax.set_xlabel("Age(years)")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(df_first_100['Age'], df_first_100['Fare'], color="darkorange", alpha=0.8)

ax.set_title("Relationship between the passenger's age and the fare")
ax.set_xlabel("Age(years)")
ax.set_ylabel("Fare($)")

plt.show()
```

```python
normal_dist = np.random.normal(loc=50, scale=100, size=1000)
gamma_dist = np.random.gamma(shape=2, scale=100, size=1000)
exp_dist = np.random.exponential(scale=100, size=1000)

fig, axs = plt.subplots(nrows=3, ncols=1, figsize=(10,8))

axs[0].hist(normal_dist, bins=30, color="tab:purple", edgecolor="black")
axs[0].set_title("Normal distribution")

axs[1].hist(gamma_dist, bins=30, color="tab:blue", edgecolor="black")
axs[1].set_title("Gamma distribution")

axs[2].hist(exp_dist, bins=30, color="tab:orange", edgecolor="black")
axs[2].set_title("Exponential distribution")

plt.tight_layout()
plt.show()
```

```python
fare_by_class = [df[df['Pclass']==cls]['Fare'].dropna() for cls in (1,2,3)]
fig, ax = plt.subplots(figsize=(10, 6))

ax.boxplot(fare_by_class, whis=[2,98], tick_labels=["1st class", "2nd class", "3rd class"], patch_artist=True, orientation="horizontal")

plt.title('Ticket Fare by Class\n(whis=[2, 98] percentiles)', fontsize=14, pad=15)
plt.ylabel('Fare (£)')
plt.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.show()
```

**Выводы**

**Анализ графиков**

1. Линейный график

   Стоимость билетов варьируется от 0 до 260$
   Присутствуют резкие скачки цен, что указывает на разные классы обслуживания

2. Гистограмма возраста пассажиров

   Большинству пассажиров 20-30 лет

3. Scatter

   В датасете нет колонок с выраженной линейной зависимостью

4. Boxplot

   Чёткая зависимость цены от класса: 1-й класс значительно дороже (медиана ~60$) чем 2-й (~14$) и 3-й (~8$)

   Аномалии:

   -В 1-м классе: выброс ~510£ (экстремально дорогой билет)
   
   -В 3-м классе: несколько билетов по 0-5£ и один выброс ~70£


5. Наблюдения по случайным распределениям

   Нормальное распределение

   -Симметричное, форма "колокола"

   Гамма распределение

   -Правосторонняя асимметрия — хвост тянется вправо

   Экспоненциальное распределение

   -Сильная асимметрия — максимум значений около 0

   -Быстрый спад с увеличением значения
