```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
documents = [
    "Робот собрал данные с поверхности Луны",
    "Бабушка испекла свежий яблочный пирог",
    "Робот срубил дерево",
    "Банк резко повысил ставку по ипотеке"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

feature_names = vectorizer.get_feature_names_out()
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names, index=[f"Doc {i}" for i in range(len(documents))])

print("TF-IDF TABLE")
display(df_tfidf.round(2))


query = ["расстояние до луны"]

query_vec = vectorizer.transform(query)

similarities = cosine_similarity(query_vec, tfidf_matrix).flatten()

print(f"\n--- RESULTS OF SEARCH: {query} ---")
ranked_indices = similarities.argsort()[::-1]
for i in ranked_indices:
    score = similarities[i]
    if score > 0:
        print(f"[Similarity: {score:.2f}] -> {documents[i]}")
```

```python
!pip install nltk
from nltk.stem.snowball import SnowballStemmer
from collections import defaultdict

stemmer = SnowballStemmer("russian")

documents = [
    "Опытные программисты написали новые программы для обучения студентов",
    "Ветеринар лечил животных, используя современные медицинские программы",
    "Студенты изучали алгоритмы и программировали на языке Python"
]

stem_index = defaultdict(set)
stem_to_original = defaultdict(set)

for doc_id, doc in enumerate(documents):
    words = doc.split()
    for word in words:
        stem = stemmer.stem(word.lower())
        stem_index[stem].add(doc_id)
        stem_to_original[stem].add(word.lower())

for stem, doc_ids in stem_index.items():
    originals = stem_to_original[stem]
    print(f"Stem '{stem}' (originals: {originals}) -> docs {doc_ids}")
        
```

```python
def search(query):
    query_stems = [stemmer.stem(word.lower()) for word in query.split()]

    found_doc_ids = set()
    matching_stems = set()

    for stem in query_stems:
        if stem in stem_index:
            found_doc_ids.update(stem_index[stem])
            matching_stems.add(stem)

    results = []
    for doc_id in sorted(found_doc_ids):
        original_doc = documents[doc_id]
        highlighted = original_doc
        for stem in matching_stems:
            for original_word in stem_to_original[stem]:
                if original_word in original_doc.lower():
                    highlighted = highlighted.replace(original_word, f"[{original_word.upper()}]")
        results.append({
            'doc_id': doc_id,
            'text': original_doc,
            'highlighted': highlighted,
            'matched_stems': matching_stems
        })

    return results


query = "новая программа"
print(f"Search results for '{query}'")
results = search(query)

for result in results:
    print(f"\nDoc {result['doc_id']}:")
    print(f"Original: {result['text']}")
    print(f"Found (highlighted): {result['highlighted']}")
    print(f"Matching stems: {result['matched_stems']}")
```
