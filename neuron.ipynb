```python
import numpy as np

def heaviside(x):
    return np.where(x>0, 1, 0)

def sigmoid(x):
    return 1/(1+np.exp(-x))

    
# class Neuron:
#     def __init__(self, num_input, activation="sigmoid"):
#         self.weights = [-2, -2]     #NAND
#         self.bias = 3

#         if activation == "heaviside":
#             self.activation_func = heaviside
#         elif activation == "sigmoid":
#             self.activation_func = sigmoid


#     def feedforward(self, X):
#         z = np.dot(self.weights, X) + self.bias
#         return self.activation_func(z)
```

```python
class Neuron:
    def __init__(self, weights, bias, activation="sigmoid"):
        self.weights = np.array(weights, dtype=float)
        self.bias = bias
        self.activation = sigmoid if activation == "sigmoid" else heaviside

    def feedforward(self, X):
        return  self.activation(np.dot(self.weights, X) + self.bias)


class Layer:
    def __init__(self, weights_matrix, biases, activation="sigmoid"):
        self.neurons = [
            Neuron(weights, bias, activation) for weights, bias in zip(weights_matrix, biases)
        ]
    def feedforward(self, X):
        return np.array([n.feedforward(X) for n in self.neurons])


hidden = Layer(
    weights_matrix=[[0.5, -0.5],
                    [0.3,  0.8],
                    [-0.6, 0.2]],
    biases=[0.1, -0.2, 0.0],
)

output = Layer(
    weights_matrix=[[0.7, -0.3, 0.5],
                    [-0.4, 0.9, 0.2]],
    biases=[0.05, -0.1],
)

X = np.array([1.0, 0.5])
h = hidden.feedforward(X)
o = output.feedforward(h)

print(f"Hidden layer: {h}")
print(f"Output layer: {o}")
```

```python
w = np.array([[0.2, -0.5,  1.0],
              [0.7,  0.1, -0.3]])
b = np.array([0.1, -0.2])
x = np.array([1.0, 2.0, 3.0])

a_loop = []
for j in range(len(w)):
    z = 0.0
    for i in range(len(x)):
        z += w[j][i] * x[i]
    z += b[j]
    a_loop.append(sigmoid(z))
a_loop = np.array(a_loop)

a_vec = sigmoid(np.dot(w, x) + b)
print(f"Loop: {a_loop}")
print(f"Vector: {a_vec}")
print("Match: ", np.allclose(a_loop, a_vec))
```
